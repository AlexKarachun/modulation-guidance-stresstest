```python
import json
import numpy as np
```

```python
with open('dataset/captions_val2014.json') as f:
    data = json.load(f)
    
```

```python
rng = np.random.default_rng(0)
ids = rng.choice(a=len(data['annotations']), size=5000, replace=False).tolist()

selected = [data['annotations'][i]['caption'].strip() for i in ids]
```

```python
selected_ided = zip(range(len(selected)), selected)
```

```python

with open('dataset/coco5000.csv', 'w',encoding='utf-8') as f:
    for idx, text in selected_ided:
        print(f'{idx}|{text}', file=f)
```
